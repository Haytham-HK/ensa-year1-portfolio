```python
# Limportaion :
import pandas as pd 
from sklearn.linear_model import LinearRegression
```

```python
# Importer data
df = pd.read_csv('student_grades.csv')
print(df.head())
```

```python
#Separer les variables et resultas
X = df[['Study_Hours']]
y = df['Grade']
```

```python
#Choisir le model et ajout de data
model = LinearRegression()
model.fit(X, y)
```

```python
#Predire pour ahmed
ahmed_hours = [[7]]
prediction = model.predict(ahmed_hours)[0]
```

```python
print("Si ahmed travaile pour 7 heurs, la prediction =",prediction)
```

```python
#comparer les vrais resultas et resultas de model
comparison = pd.DataFrame({
    'les heure de revision': df['Study_Hours'].head(5),  #prendre just les 5 premier de dataframe colone study hours
    'Vrai note':    df['Grade'].head(5),  #Meme chose pour la note
    'Model resulta': model.predict(df[['Study_Hours']].head(5))  #Prendre la prediction pour les 5 premier etudiant
})

print(comparison)

#Question: pourquoi les resultas de model ne sont pas 100% vrais
```
