```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('titanic_clean.csv')
print(df.head())
```

```python
#le model ne comprend pas les mots (homme/femmme) , en doit les traduire
df['sex'] = df['sex'].map({'male': 0, 'female': 1}) # changer male avec 0 et female avec 1 
print(df.head())
```

```python
X = df[['pclass', 'sex', 'age', 'fare']]
y = df['survived']
```

```python
model = LogisticRegression()
model.fit(X, y)
```

```python
# 4. Jack vs Rose Test
# Jack: 3rd Class, Male(0), 20 yrs, $7
jack = [[3, 0, 20, 7.5]]
# Rose: 1st Class, Female(1), 17 yrs, $100
rose = [[1, 1, 17, 100]]
```

```python
print(f"Jack : Mort ou Vivant ? {model.predict(jack)[0]} ")
print(f"rose : Mort ou Vivant ? {model.predict(rose)[0]} ")
print('0 cest mort et 1 cest vivant')
```
