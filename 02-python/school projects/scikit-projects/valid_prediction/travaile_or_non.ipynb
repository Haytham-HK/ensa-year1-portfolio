```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('job_hiring.csv')
print(df.head()) #pour niveau danglais : 0 mauvais niveau et 1 bon niveau 
```

```python
X = df[['Internships', 'English_Level']]
y = df['Status']
```

```python
model = DecisionTreeClassifier(max_depth=3)
model.fit(X, y)
```

```python
ahmed_cv = [[1, 1]]
prediction = model.predict(ahmed_cv)[0]
```

```python
print(f"Ahmed veut faire 1 stage,avec un bon niveau en anglais . Result: {prediction}")
```

```python
# 3. Visualization
plt.figure(figsize=(15, 4))

plot_tree(model, 
          feature_names=['Internships', 'English_Level'], 
          class_names=['HIRED', 'UNEMPLOYED'], 
          filled=True)

plt.title("comment le model decide (Les regles)")
plt.show()
```
