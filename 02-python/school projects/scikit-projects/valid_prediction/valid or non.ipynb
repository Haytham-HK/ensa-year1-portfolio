```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('student_pass_fail.csv')
print(df.head())
```

```python
X = df[['Grade', 'Absences']]
y = df['Result']
```

```python
model = LogisticRegression()
model.fit(X, y)
```

```python
ahmed_data = [[12, 3]] #12 note , 3 absence
prediction = model.predict(ahmed_data)[0]
```

```python
print(f"Student (Grade: 12, Absences: 3) -> Result: {prediction}")
```
